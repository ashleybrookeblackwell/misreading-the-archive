This confirms the archival hierarchy of the known test record but does not

```python
record.get("physicalOccurrences")
```

### Physical Occurrence Observation

NARA describes the physical occurrence associated with this record as
**loose sheets on paper** and identifies the **National Archives at College
Park - Textual Reference** unit as the responsible reference location.

The API separately represents:

1. the physical archival occurrence,
2. the M1947 microfilm publication, and
3. the JPG digital object.

This distinction is important because the digital image used in later OCR
evaluation is a representation of archival material rather than the physical
paper object itself.

The returned `copyStatus` value is
`Preservation-Reproduction-Reference`; its precise institutional meaning
will not be inferred beyond the metadata without additional documentation.

```python
record.get("onlineResources")
```

Together, the API metadata now distinguishes:

1. the physical archival material (`Paper`, `Loose Sheets`);
2. the M1947 microfilm context (`Roll 56`);
3. the Fold3–National Archives digitization collaboration; and
4. the JPG digital object (`0065.jpg`) exposed through the NARA Catalog.

The metadata does not, by itself, establish every technical step in the

## Experiment 0.3 — Candidate Discovery

### Purpose

Identify the first substantive archival page from M1947 Roll 56 suitable
for the pilot evaluation in *Misreading the Archive*.

Candidate selection will be based on historical and document characteristics,
not on whether existing OCR appears accurate or inaccurate.

### Candidate 001 Selection Criteria

A page is eligible if it:

- belongs to the defined M1947 Roll 56 corpus;
- contains substantive historical content rather than a target sheet,
  divider, cover page, or blank page;
- is primarily typewritten or printed text;
- is sufficiently legible for a human-verified reference transcription;
- contains multiple historically meaningful entities such as people,
  places, institutions, objects, or dates;
- contains at least one meaningful relationship among those entities;
- can be connected to enough provenance information to identify its archival
  context and digital source.

For the initial baseline, pages dominated by handwriting, severe physical
damage, complex tables, or extremely poor image quality will be reserved
for later testing rather than used as Candidate 001.

Existing OCR or transcription quality will **not** be used as a selection
criterion.

### Candidate 001 Selected

**Document:** Memorandum — *Legal ownership of archives and libraries*  
**Date:** 7 March 1947  
**Author/signatory:** Dr. Sante  
**Source collection:** M1947, Wiesbaden Central Collecting Point records  
**Fold3 image:** 231985714

#### Selection rationale

This page was selected using the candidate criteria established before
review of OCR quality.

The document is predominantly typewritten and sufficiently legible for
human verification while retaining realistic archival complexity,
including typewriter imperfections, handwritten corrections, strike-throughs,
line-break hyphenation, and a signature.

It contains numerous historically meaningful people, places, institutions,
and archival materials, as well as explicit relationships and actions among
them. These characteristics make it suitable for evaluating not only
character- and word-level OCR accuracy but also downstream preservation of
historical entities and relationships.

Existing OCR/transcription quality was not used as a selection criterion.

**Status: Candidate 001**

### Reference Transcription Protocol — Candidate 001

The reference transcription protocol was frozen **before transcription and before examination of existing OCR/transcriptions for Candidate 001**.

Key rules:

- Transcribe directly from the archival image.
- Do not consult NARA OCR, NARA contributed transcription, Fold3 text, or other existing text until the independent reference is locked.
- Preserve original spelling, capitalization, punctuation, abbreviations, and apparent typewriter errors.
- Do not silently correct historical wording or names.
- Maintain a **documentary reference** preserving handwritten additions, deletions, signatures, stamps, and other visible evidence.
- Derive a separate **OCR evaluation reference** for quantitative comparison.
- For the initial OCR baseline, evaluate the primary typewritten text separately from handwriting, signatures, stamps, and later declassification markings.
- Normalize only features such as whitespace, physical line wrapping, and clear end-of-line hyphenation when preparing CER/WER text.
- Mark uncertain or illegible text rather than guessing.
- Verify the transcription against the image in a second pass before locking it.
- Only after the reference is locked may existing OCR/transcriptions be opened for comparison.
- Any later change to the protocol must be documented rather than silently applied.

**Protocol status: Frozen before Candidate 001 transcription.**

### Protocol Amendment — Candidate Corpus Scope

**Date:** 2026-10-29

During digital-object verification, Candidate 001 was confirmed as
`0935.jpg` (object ID 210439782), located on M1947 Roll 55 rather than
Roll 56.

The initial pilot scope is therefore amended from Roll 56 alone to the
archives/libraries subsection of Entry 497 spanning Rolls 55–57.

This amendment reflects the archival organization documented in the M1947
roll list and was made before consultation of Candidate 001 OCR or contributed
transcription. It was not made in response to observed machine performance.

```python
CANDIDATE_NAID = 210439765
CANDIDATE_OBJECT_ID = 210439782

EXTRACTED_TEXT_URL = (
    f"https://catalog.archives.gov/api/v2/extractedText/"
    f"{CANDIDATE_NAID}"
)

response = client.session.get(
    EXTRACTED_TEXT_URL,
    params={"objectId": CANDIDATE_OBJECT_ID},
    timeout=30,
)

response.raise_for_status()

ocr_response = response.json()
```
